  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432074_TarinaAngelineAurelia
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import json
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432074_TarinaAngelineAurelia
torch.manual_seed(42)
np.random.seed(42)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
hasil = {}

# 1. Muat dataset
nama_file = 'Metro_Interstate_Traffic_Volume.csv'
if not os.path.exists(nama_file):
    from google.colab import files
    nama_file = list(files.upload().keys())[0]
df = pd.read_csv(nama_file)
hasil['n_rows_raw'] = int(df.shape[0])
hasil['n_cols'] = int(df.shape[1])
hasil['columns'] = list(df.columns)
print('Ukuran dataset (baris, kolom):', df.shape)
print(df.head())
print(df.isnull().sum())

# 2. Pembersihan data & pembentukan deret waktu per jam
df['date_time'] = pd.to_datetime(df['date_time'])
df = df.sort_values('date_time')
hasil['n_dupe_ts'] = int(df['date_time'].duplicated().sum())
df = df.drop_duplicates(subset='date_time', keep='first')
series = df.set_index('date_time')['traffic_volume'].resample('h').mean()
hasil['n_hours_total'] = int(len(series))
hasil['n_missing_before'] = int(series.isnull().sum())
series = series.interpolate(method='time', limit=3, limit_area='inside')
hasil['n_missing_after'] = int(series.isnull().sum())
hasil['date_start'] = str(series.index.min())
hasil['date_end'] = str(series.index.max())
hasil['vol_min'] = float(series.min())
hasil['vol_max'] = float(series.max())
hasil['vol_mean'] = float(series.mean())
hasil['vol_std'] = float(series.std())
print(series.describe())

# 3. Sliding window: 24 jam terakhir -> prediksi 1 jam berikutnya
WINDOW = 24
values = series.values.astype(float)
windows = np.lib.stride_tricks.sliding_window_view(values, WINDOW + 1)
target_time = series.index[WINDOW:]
valid = ~np.isnan(windows).any(axis=1)
windows = windows[valid]
target_time = target_time[valid]
split = int(len(windows) * 0.8)

# 4. Normalisasi Min-Max (fit hanya pada data latih)
scaler = MinMaxScaler()
scaler.fit(windows[:split].reshape(-1, 1))
scaled = scaler.transform(windows.reshape(-1, 1)).reshape(windows.shape)

# 5. Konversi ke PyTorch Tensor (X: [sampel, 24, 1], y: [sampel, 1])
X_train = torch.tensor(scaled[:split, :WINDOW], dtype=torch.float32).unsqueeze(-1)
y_train = torch.tensor(scaled[:split, WINDOW], dtype=torch.float32).unsqueeze(-1)
X_test = torch.tensor(scaled[split:, :WINDOW], dtype=torch.float32).unsqueeze(-1)
y_test = torch.tensor(scaled[split:, WINDOW], dtype=torch.float32).unsqueeze(-1)
train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)

hasil['window'] = WINDOW
hasil['batch_size'] = 64
hasil['n_windows'] = int(len(windows))
hasil['n_train'] = int(len(X_train))
hasil['n_test'] = int(len(X_test))
hasil['test_start'] = str(target_time[split])
hasil['test_end'] = str(target_time[-1])
print('Sampel latih:', X_train.shape, '| Sampel uji:', X_test.shape)
print('Rentang uji:', hasil['test_start'], 's/d', hasil['test_end'])
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432074_TarinaAngelineAurelia
class SequenceModel(nn.Module):
    def __init__(self, input_size=1, hidden_size=64, num_layers=1, output_size=1):
        super(SequenceModel, self).__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        self.fc = nn.Linear(hidden_size, output_size)

    def forward(self, x):
        out, (h_n, c_n) = self.lstm(x)
        out = out[:, -1, :]
        return self.fc(out)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432074_TarinaAngelineAurelia
epochs = 50
learning_rate = 0.001
hidden_size = 64

model = SequenceModel(hidden_size=hidden_size).to(device)
criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)
total_params = sum(p.numel() for p in model.parameters())
print('Total parameter:', total_params)

train_losses = []

for epoch in range(epochs):
    model.train()
    running = 0.0
    for xb, yb in train_loader:
        xb, yb = xb.to(device), yb.to(device)
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        running += loss.item() * xb.size(0)
    epoch_loss = running / len(train_loader.dataset)
    train_losses.append(epoch_loss)
    if (epoch + 1) % 5 == 0 or epoch == 0:
        print(f'Epoch [{epoch + 1}/{epochs}] - Loss: {epoch_loss:.6f}')

hasil['epochs'] = epochs
hasil['learning_rate'] = learning_rate
hasil['hidden_size'] = hidden_size
hasil['n_params'] = int(total_params)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432074_TarinaAngelineAurelia
# 1. Grafik penurunan loss training terhadap epoch
plt.figure(figsize=(10, 4))
plt.plot(range(1, epochs + 1), train_losses, color='#114F6A', linewidth=2, label='Training Loss')
plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE, skala ternormalisasi)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('loss_curve.png', dpi=150, bbox_inches='tight')
plt.show()

# 2. Evaluasi pada data uji
model.eval()
with torch.no_grad():
    pred_scaled = model(X_test.to(device)).cpu().numpy()
y_pred = scaler.inverse_transform(pred_scaled).flatten()
y_true = scaler.inverse_transform(y_test.numpy()).flatten()
rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
mae = float(mean_absolute_error(y_true, y_pred))
r2 = float(r2_score(y_true, y_pred))
y_persist = scaler.inverse_transform(X_test[:, -1, :].numpy()).flatten()
y_seasonal = scaler.inverse_transform(X_test[:, 0, :].numpy()).flatten()
rmse_persist = float(np.sqrt(mean_squared_error(y_true, y_persist)))
rmse_seasonal = float(np.sqrt(mean_squared_error(y_true, y_seasonal)))
print(f'RMSE LSTM          : {rmse:.2f}')
print(f'MAE  LSTM          : {mae:.2f}')
print(f'R2   LSTM          : {r2:.4f}')
print(f'RMSE persistence   : {rmse_persist:.2f}')
print(f'RMSE seasonal naive: {rmse_seasonal:.2f}')

# 3. Grafik perbandingan Data Aktual vs Prediksi (168 sampel uji pertama)
n_plot = 168
waktu_uji = target_time[split:][:n_plot]
plt.figure(figsize=(12, 5))
plt.plot(waktu_uji, y_true[:n_plot], color='#114F6A', linewidth=2, label='Data Aktual')
plt.plot(waktu_uji, y_pred[:n_plot], color='#1FA4B1', linewidth=2, linestyle='--', label='Prediksi LSTM')
plt.title('Perbandingan Data Aktual vs Prediksi LSTM (Data Uji)')
plt.xlabel('Waktu')
plt.ylabel('Traffic Volume (kendaraan/jam)')
plt.xticks(rotation=30)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.savefig('prediksi_vs_aktual.png', dpi=150, bbox_inches='tight')
plt.show()

# 4. Scatter Aktual vs Prediksi (seluruh data uji)
plt.figure(figsize=(5.5, 5.5))
plt.scatter(y_true, y_pred, s=4, alpha=0.3, color='#1FA4B1', label='Sampel uji')
plt.plot([0, y_true.max()], [0, y_true.max()], color='#114F6A', linestyle='--', label='Prediksi ideal (y = x)')
plt.title('Sebaran Prediksi vs Aktual')
plt.xlabel('Aktual (kendaraan/jam)')
plt.ylabel('Prediksi (kendaraan/jam)')
plt.legend()
plt.savefig('parity.png', dpi=150, bbox_inches='tight')
plt.show()

# 5. Simpan ringkasan hasil untuk laporan
hasil['train_losses'] = [float(l) for l in train_losses]
hasil.update({'rmse': rmse, 'mae': mae, 'r2': r2, 'rmse_persist': rmse_persist, 'rmse_seasonal': rmse_seasonal})
with open('hasil_metrik.json', 'w') as f:
    json.dump(hasil, f, indent=2)
```
